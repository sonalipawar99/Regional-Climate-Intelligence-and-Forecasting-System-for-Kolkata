```python
# Monthly Analysis
```

```python
climate_data['Month'] = climate_data['Date time'].dt.month

monthly_temperature = climate_data.groupby('Month')['Temperature'].mean()

print(monthly_temperature)
```

```python
# Monthly Temperature Graph
```

```python
monthly_temperature.plot(kind='bar', figsize=(10,5))

plt.title("Average Monthly Temperature")
plt.xlabel("Month")
plt.ylabel("Temperature")

plt.show()
```

```python
# Rainfall Anomaly Detection
```

```python
# Rain Related Conditions Filter
```

```python
rainy_days = climate_data[
    climate_data['Conditions'].str.contains(
        'Rain',
        case=False,
        na=False
    )
]

print(rainy_days.head())
```

```python
# Rainfall Anomaly Detection
# Unusual rainy days detect:
```

```python
rain_anomalies = rainy_days[
    rainy_days['Cloud Cover'] > 80
]

print(rain_anomalies)
```

```python
# Observation

## “Several days showed unusually high cloud cover with rain conditions, indicating possible rainfall anomalies.”
```

```python
# Heat Alerts System
# High Temperature Detection
```

```python
# Forecast Temperature
```

```python
from prophet import Prophet

forecast_data = climate_data[
    ['Date time', 'Temperature']
]

forecast_data.columns = ['ds', 'y']

model = Prophet()

model.fit(forecast_data)

future = model.make_future_dataframe(
    periods=30
)

forecast = model.predict(future)

model.plot(forecast)
```

```python
# Forecast Observation

# Forecast results suggest continuing seasonal temperature fluctuations over the next month.
```

```python
# Forecast Components
```

```python
model.plot_components(forecast)
```

```python
# Observation
# Forecasting results indicate continued seasonal temperature fluctuations.
```
